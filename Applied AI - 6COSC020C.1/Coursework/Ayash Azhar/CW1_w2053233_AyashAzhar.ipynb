## 1. Introduction: The Operational Bottleneck
In the domain of Human Resources and Computing, specifically within IT Service Management (ITSM), the manual processing of support tickets represents a critical operational bottleneck. Modern enterprises receive thousands of support requests daily, ranging from password resets to critical server outages. Traditionally, the incident management process is "largely manual, error prone and time consuming" (Gupta, Prasad and Mohania, 2008). A human agent must read each ticket, interpret the user's intent, and manually assign it to a resolution group.

This manual dependency creates significant delays. Research by Silva *et al.* (2018) highlights that correct categorization is the prerequisite for effective incident management; without it, tickets "bounce" between teams, increasing resolution time and costs. The goal of applying Artificial Intelligence (AI) in this domain is to automate this "Tier 1" layer, reducing the administrative burden on human staff (Li and Zhan, 2012).

### 1.1 The Business Case: The High Cost of Human Triage
The justification for automated triage is primarily financial, driven by the unsustainable costs of human turnover. Manual triage is repetitive and high-pressure work, which directly contributes to employee burnout.

* **Turnover Costs:** Recent industry data reveals that call center turnover rates average **30–45%**, with the average agent lasting just **13–15 months** in their role (Insignia Resources, 2025).
* **Financial Impact:** Replacing a single agent costs between **$10,000 and $20,000**. For a standard 100-agent team, this turnover results in an estimated loss of up to **$1.7 million annually** (Insignia Resources, 2025).

By implementing an AI agent to handle the "repetitive and emotionally taxing" task of ticket routing (a key driver of turnover), businesses can reduce agent burnout, extend tenure, and significantly lower these operational expenditures.

# Part B: Compare and Evaluate AI Techniques


**Goal:** The objective is to automate the routing, triage, and diagnosis of unstructured IT support tickets. The system must process textual input (Subject + Body) and output discrete class labels (Classification). Based on the literature review, specifically the comparative study by Prihandono and Harwahyu (2020), the three most viable techniques for evaluation are Support Vector Machines (SVM), Random Forest, and Deep Learning.

## 1. Technique 1: Support Vector Machines (SVM / SGD)
### Overview & Application
Support Vector Machines (SVM) are supervised learning algorithms that classify data by finding the optimal hyperplane in an N-dimensional space that maximizes the margin between classes . In the context of IT incident management, SVMs treat the ticket text as a "Bag of Words" vector. Prihandono and Harwahyu (2020) highlighted SVM as a top-performing candidate for IT ticket classification due to its ability to handle high-dimensional, sparse text data without overfitting.

### Data Requirements
SVMs operate on numerical vectors, typically generated via TF-IDF (Term Frequency-Inverse Document Frequency). Unlike Deep Learning, SVMs are efficient even with smaller datasets (e.g., <10,000 samples) and do not require dense word embeddings (Silva et al., 2018).

### Strengths & Weaknesses
* **Strengths:** The primary advantage is computational efficiency. SVMs are lightweight and train rapidly, making them ideal for "quick incident response" systems where low latency is critical (Nilă, Apostol and Patriciu, 2020). They are also highly explainable compared to neural networks, as the feature weights directly correlate to keyword importance.
* **Weaknesses:** SVMs lack semantic understanding. As they rely on word frequency, they often fail to capture context or sarcasm. Additionally, their performance can degrade if the dataset contains significant noise or overlapping classes, as noted in general text classification literature (Prihandono and Harwahyu, 2020).

## 2. Technique 2: Random Forest (Ensemble Learning)
### Overview & Application
Random Forest is an ensemble method that constructs a multitude of decision trees during training and outputs the class that is the mode of the classes (voting) of the individual trees . In their study of automated incident categorization, Silva et al. (2018) found that ensemble methods like Random Forest often outperform single classifiers by mitigating the variance and bias inherent in noisy IT data.

### Data Requirements
Like SVMs, Random Forest processes numerical feature vectors. However, it is generally more robust to unscaled data and does not require extensive preprocessing (like normalization) compared to Neural Networks.

### Strengths & Weaknesses
* **Strengths:** Random Forest is highly robust against overfitting. By averaging the results of hundreds of trees, it can better handle the "noisy" nature of unstructured support tickets compared to simpler linear models (Silva et al., 2018). It is also capable of capturing non-linear relationships between keywords (e.g., specific word combinations).
* **Weaknesses:** The major trade-off is inference speed. Sarnovsky and Surma (2018) note that while accurate, complex ensemble models can be slower to generate predictions than linear models, which may be a bottleneck for high-volume service desks. Furthermore, on extremely high-dimensional sparse data, Random Forests can sometimes underperform compared to linear SVMs.

## 3. Technique 3: Deep Learning (Neural Networks)
### Overview & Application
Deep Learning approaches, such as Multi-Layer Perceptrons (MLP) or Transformers, use layers of artificial neurons to learn hierarchical patterns in data

The use of sophisticated machine learning models to improve incident management models, by utilizing their ability to understand semantic context rather than just keyword presence.

### Data Requirements
Deep Learning is data-intensive. Prihandono and Harwahyu (2020) observed that while Neural Networks (MLP) achieved competitive accuracy, they required significantly more computational resources and larger labeled datasets to converge compared to SVMs.

### Strengths & Weaknesses
* **Strengths:** The defining strength is semantic accuracy. Deep Learning models can understand the meaning behind a ticket, distinguishing between "System is down" (Urgent) and "System is slow" (Performance), leading to better prioritization (Ain and Safitri, 2023).
* **Weaknesses:** The computational cost is high. These models require substantial processing power (GPUs) and memory, raising concerns regarding "Sustainability" and energy consumption. Furthermore, they function as "Black Boxes," making it difficult to interpret why a specific routing decision was made (Sarnovsky and Surma, 2018).

This project selects the SVM (SGD) approach. While Deep Learning offers potential for higher semantic understanding, the literature consistently points to SVM as the most balanced choice for this specific domain. Prihandono and Harwahyu (2020) demonstrated that SVMs provide accuracy comparable to Neural Networks for ticket classification while being significantly faster and less resource-intensive. This aligns with the project's Part E goals of Sustainability and efficiency, allowing for a lightweight, explainable prototype that operates effectively on standard hardware.

## C(b). Input Data and Pre-processing

### 1. Input Data Source & Format
The prototype utilizes a synthetic but realistic "IT Service Desk" dataset, augmented to approximately **50,000 rows** to ensure statistical significance. The key input features required for training are:
* **Subject:** A short summary of the issue (e.g., *"Login failed"*).
* **Body:** A detailed description of the problem (e.g., *"I cannot access the SAP portal since Monday..."*).

The target variables (outputs) for the supervised learning model are:
* **Type:** The target department (e.g., Request vs Incident).
* **Priority:** The urgency level (Low, Medium, High, Critical).
* **Tag_1 (Diagnosis):** The specific technical classification (e.g., Hardware, Security, Access, Bug).

### 2. Data Pre-processing Pipeline
Before the data can be fed into the Support Vector Machine (SGD) model, it undergoes a rigorous pre-processing pipeline:
* **Feature Engineering (Text Merging):** The `Subject` and `Body` columns are concatenated into a single feature `full_text`.
* **Deduplication:** Duplicate tickets were identified and removed to prevent "Data Leakage."
* **Vectorization (TF-IDF):** The raw text is transformed using **Term Frequency-Inverse Document Frequency (TF-IDF)** with an N-Gram range of `(1, 2)`. This enables the model to learn context (e.g., "server down" vs "server slow").

```python

import pandas as pd
import numpy as np
import os
import re
from google.colab import drive
from sklearn.model_selection import train_test_split

# 1. Mount Drive
drive.mount('/content/drive')

# 2. Load Data
# https://www.kaggle.com/datasets/tobiasbueck/multilingual-customer-support-tickets   dataset link
FOLDER_PATH = '/content/drive/MyDrive/AAI_CW'
FILE_NAME = 'aa_dataset-tickets-multi-lang-5-2-50-version.csv'
FULL_PATH = os.path.join(FOLDER_PATH, FILE_NAME)

print(f"Loading: {FULL_PATH}...")
df = pd.read_csv(FULL_PATH)

# 3. Data Cleaning Pipeline
print(f"Raw Row Count: {len(df)}")
df['full_text'] = df['subject'] + " " + df['body']

# Remove Duplicates
df = df.drop_duplicates(subset=['full_text'])
print(f"Rows after removing duplicates: {len(df)}")

# Remove Missing Targets (Fixes 'Length Mismatch' Error)
df_clean = df.dropna(subset=['full_text', 'type', 'priority', 'tag_1', 'language'])
print(f"Rows after removing missing values: {len(df_clean)}")

# 4. Stratified Split (80% Train / 20% Test)
X = df_clean['full_text']
y_dept = df_clean['type']       # Target 1: Routing
y_prio = df_clean['priority']   # Target 2: Triage
y_issue = df_clean['tag_1']     # Target 3: Diagnosis
y_lang = df_clean['language']   # Meta: For Bias Audit

X_train, X_test, y_d_train, y_d_test, y_p_train, y_p_test, y_i_train, y_i_test, y_l_train, y_l_test = train_test_split(
    X, y_dept, y_prio, y_issue, y_lang, test_size=0.2, random_state=42
)

print(f"\n DATA READY. Training Samples: {len(X_train)} | Testing Samples: {len(X_test)}")
```

```python
# AGENT ARCHITECTURE
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import SGDClassifier

class MultiTaskSupportClassifier:
    def __init__(self):
        # IMPROVEMENT: ngram_range=(1, 2)
        # This teaches the AI to read pairs of words like "login error" vs "human error"
        self.vectorizer = TfidfVectorizer(stop_words='english', ngram_range=(1, 2), max_features=50000)

        # 3 Independent Brains
        self.dept_classifier = SGDClassifier(loss='modified_huber', class_weight='balanced', random_state=42)
        self.prio_classifier = SGDClassifier(loss='modified_huber', class_weight='balanced', random_state=42)
        self.issue_classifier = SGDClassifier(loss='modified_huber', class_weight='balanced', random_state=42)
        self.is_trained = False

    def train(self, X_text, y_dept, y_prio, y_issue):
        print(">> Vectorizing text data (Unigrams + Bigrams)...")
        X_vectors = self.vectorizer.fit_transform(X_text)

        print(">> Training 3 Brains...")
        self.dept_classifier.fit(X_vectors, y_dept)
        self.prio_classifier.fit(X_vectors, y_prio)
        self.issue_classifier.fit(X_vectors, y_issue)

        self.is_trained = True
        print(" Training Complete.")

    def predict_batch(self, X_text):
        vectors = self.vectorizer.transform(X_text)

        # 1. Predictions
        dept_preds = self.dept_classifier.predict(vectors)
        prio_preds = self.prio_classifier.predict(vectors)
        issue_preds = self.issue_classifier.predict(vectors)

        # 2. Hard Rule (Safety Policy)
        final_prio = []
        for p, i in zip(prio_preds, issue_preds):
            if i in ['Security', 'Breach', 'Hacking']:
                final_prio.append('high')
            else:
                final_prio.append(p)

        return (dept_preds, np.array(final_prio), issue_preds)

# Initialize and Train
model = MultiTaskSupportClassifier()
model.train(X_train, y_d_train, y_p_train, y_i_train)
```

# Part D: Software Testing and Evaluation

```python
# EVALUATION & VISUALIZATION FOR PART D

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print(" Running Inference for Visualization")
y_pred_dept, y_pred_prio, y_pred_issue = model.predict_batch(X_test)

# --- 1. CONFUSION MATRICES & REPORTS ---
def plot_evaluation(y_true, y_pred, title, top_n=None):
    # Align Data
    if isinstance(y_pred, pd.Series): y_pred = y_pred.values
    y_pred_series = pd.Series(y_pred, index=X_test.index)
    common_index = y_true.index.intersection(y_pred_series.index)
    y_true = y_true.loc[common_index]
    y_pred_aligned = y_pred_series.loc[common_index]

    # Filter
    labels = y_true.unique()
    if top_n and len(labels) > top_n:
        top_labels = y_true.value_counts().head(top_n).index.tolist()
        mask = y_true.isin(top_labels)
        y_true = y_true[mask]
        y_pred_aligned = y_pred_aligned[mask]
        labels = top_labels

    labels = sorted(labels)
    cm = confusion_matrix(y_true, y_pred_aligned, labels=labels)

    # Plot Heatmap
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
    plt.title(f"{title}\nAccuracy: {accuracy_score(y_true, y_pred_aligned)*100:.2f}%")
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.show()

    # Print Report
    print(f"\n--- CLASSIFICATION REPORT: {title} ---")
    print(classification_report(y_true, y_pred_aligned, zero_division=0))
    print("="*60)

plot_evaluation(y_d_test, y_pred_dept, "Department Routing")
plot_evaluation(y_p_test, y_pred_prio, "Priority Triage")
plot_evaluation(y_i_test, y_pred_issue, "Suspected Diagnosis (Top 12)", top_n=12)


# 2. BIAS AUDIT CHART
print("\n>> Generating Bias Audit Visualization...")
audit_df = pd.DataFrame({'Language': y_l_test, 'Actual': y_d_test, 'Predicted': y_pred_dept})
major_langs = audit_df['Language'].value_counts()[lambda x: x > 50].index
audit_df = audit_df[audit_df['Language'].isin(major_langs)]

acc_by_lang = audit_df.groupby('Language').apply(
    lambda x: (x['Actual'] == x['Predicted']).mean() * 100
).sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=acc_by_lang.index, y=acc_by_lang.values, palette='viridis')
plt.title("Model Fairness Check: Accuracy by Language")
plt.ylabel("Accuracy (%)")
plt.ylim(0, 100)
plt.axhline(y=acc_by_lang.mean(), color='r', linestyle='--', label=f'Average: {acc_by_lang.mean():.1f}%')
plt.legend()
plt.show()


# 3. KEYWORD IMPORTANCE
def plot_top_words(model, target_class, n=10):
    try:
        class_index = list(model.issue_classifier.classes_).index(target_class)
        feature_names = model.vectorizer.get_feature_names_out()
        coefs = model.issue_classifier.coef_[class_index]

        # Get Top Positive (Green) and Top Negative (Red)
        top_positive_idx = np.argsort(coefs)[-n:]
        top_negative_idx = np.argsort(coefs)[:n]

        top_words = [feature_names[i] for i in top_positive_idx] + [feature_names[i] for i in top_negative_idx]
        top_scores = list(coefs[top_positive_idx]) + list(coefs[top_negative_idx])

        # Create DataFrame for easier plotting
        feat_df = pd.DataFrame({'Word': top_words, 'Score': top_scores})
        feat_df['Type'] = ['Positive' if x > 0 else 'Negative' for x in feat_df['Score']]
        feat_df = feat_df.sort_values('Score', ascending=True)

        # Plot
        plt.figure(figsize=(10, 6))
        colors = ['red' if x < 0 else 'green' for x in feat_df['Score']]
        plt.barh(feat_df['Word'], feat_df['Score'], color=colors)
        plt.title(f"What triggers (or blocks) '{target_class}'?")
        plt.xlabel("Influence Score (Green = Promotes, Red = Prevents)")
        plt.grid(axis='x', linestyle='--', alpha=0.5)
        plt.show()

    except Exception as e:
        print(f"Could not plot keywords for {target_class}: {e}")

print("\n>> Generating Enhanced Keyword Analysis...")
plot_top_words(model, 'Security')  # See what counts as a hack vs what definitely isn't
plot_top_words(model, 'Hardware')  # See what counts as broken hardware
```

```python
# Testing models with raw input
import ipywidgets as widgets
from IPython.display import display, clear_output

print("--- LIVE TEST ---")
text_input = widgets.Textarea(
    value='', placeholder='Type a ticket (e.g., "I suspect a hack")...',
    description='Ticket:', layout=widgets.Layout(width='80%', height='100px')
)
button = widgets.Button(description='Analyze', button_style='success')
output_area = widgets.Output()

def on_button_click(b):
    with output_area:
        clear_output()
        ticket = text_input.value
        if not ticket: return

        # Hybrid Prediction
        d, p, i = model.predict_batch([ticket])
        vec = model.vectorizer.transform([ticket])
        conf = np.max(model.issue_classifier.predict_proba(vec))

        print(f" TICKET: '{ticket}'")
        print("-" * 40)
        print(f" ROUTING:   {d[0]}")
        print(f" TRIAGE:    {p[0]}")
        print(f" DIAGNOSIS: {i[0]}")
        print(f" CONFIDENCE: {conf*100:.1f}%")
        print("-" * 40)

button.on_click(on_button_click)
display(text_input, button, output_area)
```
