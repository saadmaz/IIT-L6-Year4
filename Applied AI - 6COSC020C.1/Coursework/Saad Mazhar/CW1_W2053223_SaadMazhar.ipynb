## Executive Summary

In today's competitive business environment, understanding customer behaviour has become essential for delivering effective and personalised marketing campaigns. Rather than applying the same marketing strategy to every customer, organisations increasingly rely on Artificial Intelligence (AI) to identify meaningful customer groups and make data-driven marketing decisions.

By identifying customers with similar attributes, the system enables businesses to develop targeted marketing strategies that improve customer engagement, strengthen customer relationships, and optimise marketing resources.

# Part A - Literature Review

## A.1 Artificial Intelligence in Marketing

Artificial Intelligence (AI) has transformed modern marketing by enabling organisations to make informed decisions using large volumes of customer data. Unlike traditional marketing approaches, which often rely on broad demographic assumptions and mass communication, AI allows businesses to analyse customer behaviour, identify hidden patterns, and deliver personalised experiences at scale. As companies collect increasing amounts of transactional and behavioural data, AI has become an essential tool for improving customer engagement, increasing operational efficiency, and enhancing decision-making (Davenport, Guha, Grewal & Bressgott, 2020).

Machine learning techniques have become particularly valuable in marketing because they can process complex datasets that would be difficult to analyse manually. These techniques enable businesses to predict customer behaviour, recommend relevant products, optimise marketing campaigns, and improve customer retention. Rather than replacing marketers, AI supports decision-making by providing insights that allow organisations to develop more effective customer strategies (Huang & Rust, 2021).

## A.2 Customer Segmentation

Customer segmentation is the process of dividing customers into groups that share similar characteristics, behaviours, or purchasing patterns. The primary objective is to enable businesses to understand the diverse needs of their customers and deliver marketing campaigns that are more relevant to each group. Traditional segmentation methods commonly relied on demographic variables such as age, gender, or income. While useful, these approaches often fail to capture the complexity of modern consumer behaviour (Wedel & Kamakura, 2012).

Recent advances in machine learning have shifted customer segmentation towards data-driven approaches that automatically discover meaningful customer groups from historical data. Instead of relying solely on predefined rules, unsupervised learning algorithms identify similarities between customers based on purchasing behaviour, spending habits, campaign responses, and engagement levels. This enables businesses to create more accurate customer profiles and develop targeted marketing strategies that improve customer satisfaction and marketing effectiveness (Xu & Wunsch, 2009).

## A.3 Machine Learning Techniques for Customer Segmentation

Customer segmentation is generally considered an unsupervised machine learning problem because customer groups are not known before analysis. Unlike supervised learning, which requires labelled training data, clustering algorithms identify natural groupings within the dataset without predefined categories.

Among the most widely used clustering algorithms is K-Means clustering, which partitions observations into a predefined number of clusters by minimising the distance between customers and their assigned cluster centroids. K-Means is widely adopted because of its computational efficiency, simplicity, and ability to perform well on large datasets (Jain, 2010).

Alternative clustering methods include Hierarchical Clustering and DBSCAN. Hierarchical Clustering creates a tree-like structure that illustrates relationships between customer groups, making it useful for exploratory analysis. DBSCAN identifies clusters based on data density and is particularly effective when clusters have irregular shapes or when datasets contain significant numbers of outliers. Although each algorithm has advantages, K-Means remains one of the most widely applied techniques in customer analytics due to its scalability and ease of interpretation (Han, Kamber & Pei, 2012).

## A.4 Personalisation and Marketing Automation

Personalisation has become one of the defining characteristics of modern digital marketing. Customers increasingly expect businesses to deliver products, services, and promotional content that reflect their individual interests and purchasing behaviour. AI-powered customer segmentation enables organisations to move beyond generic advertising by tailoring marketing campaigns to specific customer groups.

Marketing automation platforms further extend these capabilities by automatically delivering personalised emails, promotional offers, loyalty rewards, and product recommendations based on customer segments. This improves customer engagement while reducing the time and effort required for manual campaign management. Consequently, AI-driven segmentation has become an important component of customer relationship management (CRM) systems and digital marketing platforms (Chaffey & Ellis-Chadwick, 2022).

## A.5 Challenges and Ethical Considerations

Despite its advantages, AI-driven customer segmentation presents several technical and ethical challenges. One of the main limitations of clustering algorithms is determining the appropriate number of customer segments, as different choices may produce different marketing strategies. Additionally, clustering algorithms are highly sensitive to feature selection and data preprocessing, making careful data preparation essential for meaningful results.

From an ethical perspective, organisations must ensure that customer segmentation does not unintentionally reinforce bias or discriminate against specific groups. Variables such as income, location, or purchasing behaviour may indirectly act as proxies for protected characteristics, resulting in unfair marketing practices. Businesses must therefore balance commercial objectives with fairness, transparency, and responsible data governance. Compliance with privacy regulations such as the General Data Protection Regulation (GDPR) also requires organisations to collect and process customer data responsibly (European Parliament & Council, 2016).

## A.6 Future Trends

The future of AI in marketing is expected to involve increasingly intelligent and adaptive customer analytics. Advances in real-time data processing, predictive analytics, and generative AI are enabling businesses to personalise customer experiences with greater accuracy than ever before. AI systems are also becoming more explainable, allowing organisations to better understand and justify automated marketing decisions.

Another emerging trend is the integration of customer segmentation with generative AI systems capable of automatically producing personalised marketing content for different customer groups. As organisations continue investing in AI technologies, customer segmentation is likely to evolve from periodic analysis into continuous, real-time decision-making that supports highly personalised customer experiences while maintaining privacy, fairness, and sustainability.

# Part B - Compare and Evaluate Three AI Techniques

**Goal:** segment customers in the Customer Personality Analysis dataset into a small,
explainable number of groups so that Marketing can design a distinct campaign/action per
group, rather than treating every customer identically. Three unsupervised clustering
techniques are compared below; K-Means is then selected for in-depth implementation in
Part C.

## B.1 K-Means (chosen for Part C)

**What it is:** a centroid-based partitional algorithm. It iteratively assigns each point
to its nearest of K centroids, then recomputes each centroid as the mean of its assigned
points, minimising within-cluster variance (Jain, 2010).

**Strengths:** computationally efficient (O(n·k·i)) and scales comfortably to the
thousands-of-customers range typical of CRM systems; output is directly interpretable,
since each centroid is a "typical customer" profile a marketer can read off directly; has
only one real hyperparameter (K), which can be tuned quantitatively via the elbow method
and silhouette score rather than guesswork.

**Weaknesses:** K must be chosen in advance; assumes clusters are roughly spherical and
similarly sized, which real customer behaviour does not always respect; every point is
forced into a cluster, so it cannot flag outliers on its own; sensitive to feature scaling
and initial centroid placement (mitigated here via `n_init=10` restarts).

**Example application:** partitioning customers on (Income, Total_Spending, Recency,
Total_Purchases) into segments such as "high-value, campaign-responsive" versus
"low-engagement, price-sensitive," each mapped to a distinct marketing action.

**Data, setup time, runtime, output:** requires a numeric, scaled feature matrix; setup is
fast (few hyperparameters beyond K); fitting ~2,200 rows completes in well under a second;
output is a hard cluster label per customer plus centroid coordinates for interpretation.

## B.2 Hierarchical Clustering

**What it is:** builds a tree (dendrogram) of nested clusters, typically agglomeratively
(bottom-up, repeatedly merging the closest pair of clusters), without requiring K to be
fixed in advance (Han, Kamber & Pei, 2012).

**Strengths:** K is chosen post-hoc by cutting the dendrogram at a chosen height, so the
number of segments doesn't have to be committed to before seeing the data; the dendrogram
itself is a useful exploratory visualisation, showing sub-segment structure at multiple
levels of granularity; depending on linkage method, can capture more flexible cluster
shapes than K-Means.

**Weaknesses:** O(n²) time/memory complexity or worse, which becomes impractical well
before the row counts common in real CRM databases; results are sensitive to the choice of
linkage method (single/complete/average/Ward) and distance metric; still fundamentally a
proximity-based method, so it shares some of K-Means' shape assumptions depending on
linkage.

**Example application:** could reveal whether "high-value" customers actually split into
finer sub-segments (e.g. wine-focused versus general-goods high spenders) before deciding
how many distinct campaigns to commission.

**Data, setup time, runtime, output:** same scaled numeric input as K-Means; setup requires
choosing a linkage method upfront; runtime degrades sharply as row count grows, making it
impractical much beyond a few thousand rows without approximation; output is a dendrogram
plus cluster labels once a cut height is chosen.

## B.3 DBSCAN

**What it is:** a density-based algorithm that groups closely-packed points together and
marks points in low-density regions as noise/outliers, controlled by two parameters:
`eps` (neighbourhood radius) and `min_samples` (Han, Kamber & Pei, 2012).

**Strengths:** does not require K to be specified upfront; can discover arbitrarily-shaped
clusters, not just convex/spherical ones; naturally separates outliers from the main
clusters instead of forcing every point into a segment, which could help flag anomalous
purchasing behaviour.

**Weaknesses:** very sensitive to the `eps`/`min_samples` parameters, which are difficult to
tune reliably on customer data; struggles when clusters have noticeably different
densities, which is common here, since high-value customers are naturally sparser than average
ones; not designed to guarantee a fixed, marketing-friendly number of segments, which is a
practical business constraint in this domain (see justification below).

**Example application:** could be used to flag anomalous customers (e.g. possible
data-entry errors or unusual behaviour) as noise, as a complement to the main segmentation
rather than a replacement for it.

**Data, setup time, runtime, output:** same scaled numeric input; setup time is higher,
since `eps`/`min_samples` typically require a k-distance plot or grid search to tune;
runtime is moderate (roughly O(n log n) with spatial indexing); output is a cluster label
per point, including a distinct "noise" label (−1) for points that don't belong to any
cluster.

## Comparison Summary

| Criterion | **K-Means** (chosen) | Hierarchical Clustering | DBSCAN |
|---|---|---|---|
| Requires K upfront | Yes | No (cut dendrogram post-hoc) | No |
| Scalability | Good (O(n·k·i)) | Poor for large n (O(n²) or worse) | Moderate |
| Cluster shape assumption | Spherical, similar variance | Flexible (depends on linkage) | Arbitrary shape |
| Handles outliers | Poorly (outliers pull centroids) | Poorly | Well (labels outliers as noise) |
| Interpretability | High (centroid = "typical customer") | High (dendrogram shows hierarchy) | Moderate |
| Sensitive to parameter choice | K only | Linkage method only | Very sensitive (eps, min_samples) |
| Typical dataset size fit | Large | Small–medium | Small–medium, uneven density |
| Business use case fit | Segment counts are usually a marketing decision anyway (e.g. "we can only run 4 campaigns") | Useful for exploratory hierarchy of sub-segments | Useful for anomaly/fraud-style detection, less so for marketing segments |

**Why K-Means was selected for Part C:** marketing segmentation almost always needs a
small, fixed, *explainable* number of groups, because the real business constraint is "how
many distinct campaigns can we realistically design and staff," not "however many natural
clusters the data contains." K-Means' centroid-based output maps directly onto that
constraint, scales comfortably to the dataset size used here, and its main weakness
(sensitivity to K) is directly addressed in the Testing & Evaluation section via the elbow
method and three independent validity indices, whereas DBSCAN's variable, data-driven
cluster count and Hierarchical Clustering's poor scalability are both a worse fit for this
specific business constraint.

## C.0 (continued) Input Data & Pre-processing

**Source and format.** The input is the public Kaggle "Customer Personality Analysis"
dataset (`imakash3011/customer-personality-analysis`), embedded directly in the
notebook as a tab-separated `CSV_DATA` string so it runs without any external file or
network access. It contains 2,240 rows (one per customer) and 29 columns,
well above the >1,000-row guideline for an ML algorithm. Columns fall into three groups:
demographics (`Year_Birth`, `Education`, `Marital_Status`, `Income`, `Kidhome`,
`Teenhome`), transaction history (six product-category spend columns `MntWines`…
`MntGoldProds`, and five purchase-channel columns `NumDealsPurchases`…
`NumWebVisitsMonth`), and campaign-response flags (`AcceptedCmp1`–`AcceptedCmp5`,
`Response`, `Complain`). `Dt_Customer` (enrolment date) is stored as `DD-MM-YYYY` text and
must be parsed explicitly: `pandas.read_csv` does not infer this format automatically.

**Pre-processing pipeline (Part C, Steps 1–2):**
1. **Uninformative-column removal**: `Z_CostContact` and `Z_Revenue` are constant
   across every row (always 3 and 11) and carry no information; `Complain` is not
   constant but is not used by any hypothesis in this analysis. All three are dropped
   immediately after loading.
2. **Missing-value handling**: `Income` has 24 missing values (~1% of rows); these rows
   are dropped rather than imputed, so every downstream step uses complete records only.
3. **Feature engineering**: raw columns are combined into RFM-style features:
   `Total_Spending` (sum of the six spend columns), `Total_Purchases` (sum of four
   purchase-channel columns), `Total_Campaigns_Accepted` (sum of six campaign-response
   flags), `Children` (`Kidhome` + `Teenhome`), `Customer_Tenure_Days` (days since
   enrolment), and `Age` (derived from `Year_Birth`; implausible values above 90 are
   dropped).
4. **Outlier handling**: `Income` and `Total_Spending` are capped at the 1st/99th
   percentile rather than dropped, limiting the influence of extreme values on K-Means
   centroids (see Part B) while preserving sample size.
5. **Feature selection & scaling**: clustering uses a deliberately small, interpretable
   core feature set (`Income`, `Total_Spending`, `Recency`, `Total_Purchases`),
   standardised with `StandardScaler` (zero mean, unit variance) so no single feature
   dominates the Euclidean distance calculation purely due to its numeric scale.

### Step 1: Load & Explore Data

Loaded directly from the `CSV_DATA` string embedded in the notebook above (the real
Customer Personality Analysis dataset, tab-separated), so the notebook is fully
self-contained and runs without any external file or network dependency. Rows with
missing values are dropped immediately, so everything below uses complete data only.

````python
print("STEP 1: Load and explore data")

df = pd.read_csv(io.StringIO(CSV_DATA), sep="	")
print(f"Loaded embedded dataset: {df.shape}")

# Z_CostContact/Z_Revenue are constant for every row (always 3 and 11) so carry no
# information; Complain is not constant but is unused anywhere in this analysis
df = df.drop(columns=["Z_CostContact", "Z_Revenue", "Complain"])

# pd.read_csv doesn't auto-parse dates; the real dataset's Dt_Customer is DD-MM-YYYY text
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)

# Drop rows with missing values immediately, so all exploration below uses complete data only
before = len(df)
df = df.dropna().copy()
print(f"Dropped {before - len(df)} rows with missing values; {len(df)} rows remain")

df.info()
print("\nMissing values:\n", df.isnull().sum()[df.isnull().sum() > 0])

plt.figure(figsize=(8, 3))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
plt.title("Missing Value Map")
plt.tight_layout()
plt.show()

df.describe(include="all").T
````

### Step 2: Clean & Prepare Data

Engineer RFM-style features and cap outliers, then scale a small, easy-to-explain core
feature set (Income, Total Spending, Recency, Total Purchases) for clustering (missing
values were already dropped in Step 1).

````python
print("STEP 2: Clean, engineer features, and prepare for clustering")

# Feature engineering
df["Age"] = pd.Timestamp("2015-01-01").year - df["Year_Birth"]
df = df[df["Age"] <= 90].copy()  # drop a handful of implausible ages

spend_cols = ["MntWines", "MntFruits", "MntMeatProducts",
              "MntFishProducts", "MntSweetProducts", "MntGoldProds"]
df["Total_Spending"] = df[spend_cols].sum(axis=1)

purchase_cols = ["NumDealsPurchases", "NumWebPurchases", "NumCatalogPurchases", "NumStorePurchases"]
df["Total_Purchases"] = df[purchase_cols].sum(axis=1)

campaign_cols = ["AcceptedCmp1", "AcceptedCmp2", "AcceptedCmp3", "AcceptedCmp4", "AcceptedCmp5", "Response"]
df["Total_Campaigns_Accepted"] = df[campaign_cols].sum(axis=1)

df["Children"] = df["Kidhome"] + df["Teenhome"]
df["Customer_Tenure_Days"] = (pd.Timestamp("2015-01-01") - df["Dt_Customer"]).dt.days
df["Marital_Status"] = df["Marital_Status"].replace(
    {"Alone": "Single", "Absurd": "Single", "YOLO": "Single"}
)

# Cap outliers rather than dropping rows, to preserve sample size (K-Means is outlier-sensitive)
for col in ["Income", "Total_Spending"]:
    lower, upper = df[col].quantile([0.01, 0.99])
    df[col] = df[col].clip(lower, upper)

# Correlation heatmap + key distributions
numeric_cols = df.select_dtypes(include=[np.number]).columns.drop("ID")
plt.figure(figsize=(11, 9))
sns.heatmap(df[numeric_cols].corr(), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df["Age"], bins=30, ax=axes[0], kde=True)
axes[0].set_title("Age Distribution")
sns.histplot(df["Income"], bins=30, ax=axes[1], kde=True)
axes[1].set_title("Income Distribution")
sns.histplot(df["Total_Spending"], bins=30, ax=axes[2], kde=True)
axes[2].set_title("Total Spending Distribution")
plt.tight_layout()
plt.show()

# Scale a small, easy-to-explain core feature set for clustering (classic RFM + income)
cluster_features = ["Income", "Total_Spending", "Recency", "Total_Purchases"]
X = df[cluster_features].copy()
X_scaled = pd.DataFrame(StandardScaler().fit_transform(X), columns=cluster_features, index=X.index)

print(f"Prepared {len(df)} rows x {len(cluster_features)} clustering features: {cluster_features}")
X_scaled.describe().T
````

### Step 3: Build the Clustering Model

Compare K=2–10 via the elbow method and silhouette score, then fit and visualize the final
K-Means model. `FINAL_K` is set explicitly; justify the value in the Testing & Evaluation
discussion, citing the actual numbers below.

````python
print("STEP 3: Choose K and fit the final clustering model")

# Compare K=2-10 on inertia (elbow) and silhouette score
k_range = range(2, 11)
inertias, sil_scores = [], []
for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(list(k_range), inertias, marker="o")
axes[0].set_xlabel("Number of clusters (K)")
axes[0].set_ylabel("Inertia")
axes[0].set_title("Elbow Method")
axes[1].plot(list(k_range), sil_scores, marker="o", color="darkorange")
axes[1].set_xlabel("Number of clusters (K)")
axes[1].set_ylabel("Silhouette Score")
axes[1].set_title("Silhouette Score vs K")
plt.tight_layout()
plt.show()

print("K with highest silhouette score:", list(k_range)[int(np.argmax(sil_scores))])

# Fit the final model
FINAL_K = 4  # <-- justify this choice in the Testing & Evaluation discussion before submitting
kmeans_final = KMeans(n_clusters=FINAL_K, random_state=RANDOM_STATE, n_init=10)
df["Cluster"] = kmeans_final.fit_predict(X_scaled)

pca = PCA(n_components=2, random_state=RANDOM_STATE)
df["PCA1"], df["PCA2"] = pca.fit_transform(X_scaled).T

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x="PCA1", y="PCA2", hue="Cluster", palette="Set2", alpha=0.75, ax=axes[0])
axes[0].set_title(f"Customer Segments - PCA Projection (K={FINAL_K})")
df["Cluster"].value_counts().sort_index().plot(kind="bar", color=sns.color_palette("Set2"), ax=axes[1])
axes[1].set_xlabel("Cluster")
axes[1].set_ylabel("Number of customers")
axes[1].set_title("Cluster Sizes")
plt.tight_layout()
plt.show()

print("Cluster sizes:\n", df["Cluster"].value_counts().sort_index())
````

### Step 4: Profile Segments & Recommend Actions

Summarize each cluster, visualize the profiles, then auto-generate a plain-language
description and marketing action per segment; read these against your own judgement
before writing them up.

````python
print("STEP 4: Profile segments and generate marketing recommendations")

profile_cols = ["Age", "Income", "Total_Spending", "Total_Purchases",
                "Total_Campaigns_Accepted", "Recency", "Children", "NumWebVisitsMonth",
                "Customer_Tenure_Days"]
cluster_profile = df.groupby("Cluster")[profile_cols].mean().round(1)
print(cluster_profile)

# Radar chart of normalised cluster profiles
radar_norm = (cluster_profile - cluster_profile.min()) / (cluster_profile.max() - cluster_profile.min())
categories = list(radar_norm.columns)
angles = [n / float(len(categories)) * 2 * pi for n in range(len(categories))]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
for cluster_id in radar_norm.index:
    values = radar_norm.loc[cluster_id].tolist()
    values += values[:1]
    ax.plot(angles, values, label=f"Cluster {cluster_id}")
    ax.fill(angles, values, alpha=0.08)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=8)
ax.set_title("Cluster Profiles (Normalised)")
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.show()

# Spending & income by cluster
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x="Cluster", y="Total_Spending", palette="Set2", ax=axes[0])
axes[0].set_title("Total Spending by Cluster")
sns.boxplot(data=df, x="Cluster", y="Income", palette="Set2", ax=axes[1])
axes[1].set_title("Income by Cluster")
plt.tight_layout()
plt.show()

# Rule-based cluster interpretation (auto-generated; read against your own domain judgement)
def describe_cluster(row, overall):
    tags = []
    if row["Income"] > overall["Income"] * 1.15:
        tags.append("high income")
    elif row["Income"] < overall["Income"] * 0.85:
        tags.append("low income")
    if row["Total_Spending"] > overall["Total_Spending"] * 1.15:
        tags.append("high spender")
    elif row["Total_Spending"] < overall["Total_Spending"] * 0.85:
        tags.append("low spender")
    if row["Total_Campaigns_Accepted"] > overall["Total_Campaigns_Accepted"] * 1.15:
        tags.append("campaign-responsive")
    if row["Recency"] > overall["Recency"] * 1.15:
        tags.append("lapsing / inactive")
    if row["NumWebVisitsMonth"] > overall["NumWebVisitsMonth"] * 1.15:
        tags.append("web-heavy browser, lower conversion")
    return ", ".join(tags) if tags else "near-average on all measured features"

overall_avg = cluster_profile.mean()
print("\nCluster interpretation:")
for cluster_id, row in cluster_profile.iterrows():
    print(f"Cluster {cluster_id}: {describe_cluster(row, overall_avg)}")

# Rule-based marketing recommendation engine
def recommend_action(row, overall):
    high_value = row["Income"] > overall["Income"] and row["Total_Spending"] > overall["Total_Spending"]
    campaign_responsive = row["Total_Campaigns_Accepted"] > overall["Total_Campaigns_Accepted"] * 1.15

    if high_value and campaign_responsive:
        return "VIP / loyalty programme: exclusive early access, premium bundles"
    if high_value and not campaign_responsive:
        return ("High-value, campaign-unresponsive: switch to personalised outreach "
                "(e.g. dedicated account contact) rather than more generic campaigns")
    if row["Recency"] > overall["Recency"] * 1.2:
        return "Win-back campaign: reactivation offer, personalised outreach"
    if row["Total_Spending"] < overall["Total_Spending"] * 0.8:
        return "Discount / deals-led campaign: coupons, bundle pricing"
    if row["Customer_Tenure_Days"] < overall["Customer_Tenure_Days"] * 0.7:
        return "Onboarding journey: welcome sequence, first-purchase incentive"
    return "Standard nurture campaign: periodic newsletter, no discount needed"

recommendations = pd.DataFrame({
    "Cluster": cluster_profile.index,
    "Segment Profile": [describe_cluster(r, overall_avg) for _, r in cluster_profile.iterrows()],
    "Recommended Action": [recommend_action(r, overall_avg) for _, r in cluster_profile.iterrows()],
    "Cluster Size": df["Cluster"].value_counts().sort_index().values,
})
recommendations
````

````python
labels_final = df["Cluster"]
sil = silhouette_score(X_scaled, labels_final)
db = davies_bouldin_score(X_scaled, labels_final)
ch = calinski_harabasz_score(X_scaled, labels_final)

print(f"Silhouette Score:        {sil:.3f}  (higher is better, range -1 to 1)")
print(f"Davies-Bouldin Index:    {db:.3f}  (lower is better)")
print(f"Calinski-Harabasz Score: {ch:.1f}  (higher is better)")
````

### Comparison Across K Values

````python
comparison_rows = []
for k in [3, 4, 5, 6]:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    lbl = km.fit_predict(X_scaled)
    comparison_rows.append({
        "K": k,
        "Silhouette": round(silhouette_score(X_scaled, lbl), 3),
        "Davies-Bouldin": round(davies_bouldin_score(X_scaled, lbl), 3),
        "Calinski-Harabasz": round(calinski_harabasz_score(X_scaled, lbl), 1),
    })
pd.DataFrame(comparison_rows)
````

### Discussion

Different values of **K** were evaluated to determine the most appropriate number of customer segments (see comparison table above). While **K = 2** achieved the highest overall Silhouette Score, it produced only two broad customer groups, too coarse to support distinct marketing actions. Among **K = 3–6**, **K = 3** scored best on all three validity indices (Silhouette 0.355, Davies–Bouldin 0.998, Calinski–Harabasz 1722.0), moderately ahead of **K = 4** (0.318, 1.202, 1657.9).

Despite K = 3 scoring higher on these purely statistical measures, **K = 4** was retained as the final model for a business reason. Comparing the Step 4 cluster profiles shows that K = 4 splits each income/spending tier into an actively engaged sub-segment and a lapsing one: Clusters 0 and 2 are both high-income, high-spending, campaign-responsive customers, but Cluster 2 is additionally flagged as lapsing/inactive; the same pattern separates Cluster 1 from Cluster 3 among lower-value customers. At K = 3 this distinction collapses, so a lapsing high-value customer would receive the same loyalty-programme treatment as an actively engaged one, even though the recommended action for each (win-back offer versus loyalty rewards) is materially different. Given that the quality gap between K = 3 and K = 4 is modest — both fall in the weak-to-moderate range typical of real behavioural data, rather than one being a clearly correct clustering and the other wrong — the small drop in aggregate cluster-quality metrics is an acceptable trade-off for a segmentation that distinguishes engaged from lapsing customers within both value tiers, which is directly actionable for campaign design. This mirrors the justification in Part B: the number of segments is ultimately a business constraint on how many distinct campaigns can be designed and resourced, not solely a function of which K maximises a single clustering-quality index.

### Limitations

* K-Means assumes clusters are similar in shape and size, which may not accurately represent real customer behaviour.
* Customer segments may change over time as purchasing patterns evolve.
* The model uses a small set of core features, which improves interpretability but may overlook other factors that influence customer behaviour.

### Possible Improvements

* Compare the results with other clustering techniques such as DBSCAN or Gaussian Mixture Models.
* Test the stability of the clusters using different samples of the dataset.
* Evaluate the model using more recent or time-based customer data to improve its long-term reliability.

# Part E - Equity, Diversity, Inclusion (EDI) and Sustainability

**Bias and fairness**
Income- and age-based features are strong clustering signals precisely because they
correlate with protected characteristics. A cluster that looks purely behavioural
("low spenders") can still function as a proxy for a protected group even when that
attribute was never fed into the model directly. Any deployed version of this pipeline
needs a proxy-discrimination check, not just a check on the input feature list.

**Privacy / GDPR**
Purchase history, income, and web-visit counts are personal data under GDPR. A production
system needs a documented lawful basis for processing, a defined retention period for
`Recency`/`Dt_Customer`-derived features, and a mechanism for a customer to request removal
from a segment: "the model already trained on it" is not currently an accepted GDPR
exemption.

**Explainability**
Marketers acting on a segment need to be able to explain, in plain language, why a given
customer landed in that segment. The rule-based `describe_cluster` function above is a
minimal step toward this; a production system would need this validated against real
marketer judgement, not just numeric thresholds.

**Sustainability**
K-Means is computationally cheap relative to deep learning approaches: training here
completes in seconds even on typical dataset sizes. The larger sustainability cost is
retraining cadence: retraining on every new transaction is unnecessary; a scheduled
(e.g. monthly) batch retrain is very likely sufficient and meaningfully reduces compute cost
compared with continuous retraining. `[Likely; verify against your own runtime figures on
the real dataset rather than asserting this without measurement.]`
